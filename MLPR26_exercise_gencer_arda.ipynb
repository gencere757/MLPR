```python
#Importing Required Packages For the Whole Project
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from skimage import io, img_as_float, color
from skimage.feature import graycomatrix, graycoprops
from skimage.transform import resize
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import cv2
```

```python
#Reading the Images
DATASET_PATH = os.path.join(".", "Dataset") #The dataset path


#Loads the data and creates a dictionary where keys are labels and values are lists of images
def load_data(datasetPath):
    print(f"Loading data paths:")
    dataPaths = dict()   #Keys are classes, values are lists of paths of images
    for elem in os.listdir(datasetPath):   #Go into each class folder
        path = os.path.join(datasetPath, elem) #Get folder path
        print(f"Walking into {path}")
        imagePaths = list()
        for image in os.listdir(path):  #Go into each image path inside class folder
            imagePaths.append(os.path.join(path, image))    #Add image to path list
        dataPaths[elem] = imagePaths #Add list to dictionary
    #Go into directories and get images as actual image objects
    datasets = dict()   
    for label, paths in dataPaths.items():
        datasets[label] = list()
        for image in paths:
            img = img_as_float(io.imread(image))
            datasets[label].append(img)
    return datasets

datasets = load_data(DATASET_PATH)
```

```python
#Formatting the  data
#Converts images to grayscale in the whole dataset
def convertToGray(dataset):
    greyedImgs = dict()
    for label, imgs in dataset.items():
        greyedImgs[label] = list()
        for img in imgs:
            greyedImgs[label].append(color.rgb2gray(img))
    return greyedImgs

#Changes the quantization level of all images in dataset, assumes initial quantization of 256 levels
def changeQuantization(dataset, qLevel):
    for label, data in dataset.items():
        for idx in range(len(data)):
            img = data[idx] #Get original image
            img = np.array(img) #Convert to array
            img = np.floor(img * qLevel).astype(np.uint8)    #Apply new quantization level and convert to integer
            data[idx] = img #Update list
        dataset[label] = data   #Update the key
    return dataset

modifiedDatasets = convertToGray(datasets)  #Convert to greyscale
modifiedDatasets = changeQuantization(modifiedDatasets,8)   #Reduce quantization to 8 gray levels
#Testing to see some images
for label, images in modifiedDatasets.items():
    for img in images:
        plt.imshow(img, cmap='gray')
        plt.title("An example formatted image")
```

### First Order Texture Measures 

```python
#Extracts first level features for a singular image
def getfeaturesDict(img):
    #Get each color channel
    R = img[..., 0]
    G = img[..., 1]
    B = img[..., 2]

    #Get means of each color channel
    meanR = np.mean(R)
    meanG = np.mean(G)
    meanB = np.mean(B)

    #Get variances of each color channel
    varR = np.var(R)
    varG = np.var(G)
    varB = np.var(B)

    return {
        "meanR": meanR,
        "meanG": meanG,
        "meanB": meanB,
        "varR": varR,
        "varG": varG,
        "varB": varB,
    }

#Extracts a single feature and returns as np array
def extractFeature(featureDataset, feature, oneClass = "no"):
    listToStore = list()
    if oneClass != "no":
        images = featureDataset[oneClass]
        for features in images:
            listToStore.append(features[feature])
    else:
        for label, images in featureDataset.items():
            for features in images:
                listToStore.append(features[feature])
    return np.array(listToStore)

#Calculate first order measures for each image
featuresDict = dict()
for label, images in datasets.items():
    temp = list()
    for img in images:
        features = getfeaturesDict(img)
        features["class"] = label   #Add the label to the features as well
        temp.append(features)
    featuresDict[label] = np.array(temp)
```

### Second Order Texture Measures

```python
#The distances and angles for which to calculate GLCM matrix
distancesGLCM = [1,2]
anglesGLCM = [0,90]
#Go through the greyscale images and extract their correlation features
for label, images in modifiedDatasets.items():
    temp = list()
    idx  = 0
    for img in images:
        glcm = graycomatrix(image=img,levels=8,distances=distancesGLCM,angles=anglesGLCM,symmetric=True)   #Compute GLCM for 3 distances and 3 angle values
        corr = graycoprops(glcm,prop='correlation')
        featuresDict[label][idx]["correlation"] = corr
        idx += 1
```

Correlation feature is a second order texture measure for greyscale images that is calculated by utilizing the GLCM. It is calculated by summing up the normalized elements of a GLCM for a given distance andd angle (in a similar fashion for calculating pearson correlation for a given pair of gray levels.). In short, it measures how smooth an image is. If the correlation is close to 1, it means that we are much more likely to find a gray level j that is close or the same as another gray level i in an image. This would imply that the image is smooth and there isn't much steep changes in intensity levels in the image. Similarly, a correlation of 0 would imply there is no correlation between the two given gray levels in the image, meaning that the gray levels are quite random in the image. And finally, a negative correlation implies that the gray levels are inversely proportional to each other, meaning that you are much more likely to find stark contrasts in the image.

### Template Matching Features

```python
#Template matching
TEMPLATE_SIZE = (50, 50)
templates = {} #The  dict to store the template images for each class

for label in modifiedDatasets.keys():
    # Take the first image of each class as template
    raw_template = modifiedDatasets[label][0]
    # Resize to a standard small size
    templates[label] = cv2.resize(raw_template, TEMPLATE_SIZE)

#The loops iterate through each class and each image inside that class
for label, imagesList in modifiedDatasets.items():
    for idx, gray in enumerate(imagesList):
        
        #This part of the code was taken from the exercise slides
        template_scores = list() 
        for t_name, t_img in templates.items(): 
            t_resized = cv2.resize(t_img, (gray.shape[1], gray.shape[0]))
            res = cv2.matchTemplate(gray, t_resized, cv2.TM_CCOEFF_NORMED)
            template_scores.append(res.max())
            
            #Store result in the features dictionary
            featuresDict[label][idx][f"tmatch{t_name.capitalize()}"] = res.max()
```

The difference between first and second order measures and template measures are that the previous features are essentially statistical features, meaning that they only focus on quantitative properties (first order measures) or small scale spatial relationships between pixels (that are quite near each other) about the image such as number of a certain color or number of co-occuring pixels of given gray values with a certain distance & angle in an image. Template matching, on the other hand, focuses on also the certain structures or how pixels or patches are oriented in an image as well on a much larger scale than GLCM. This is done by selecting an representative image from each class and comparing the characteristics of other images with this selected representative. This approach also allows to get the overall shape etc. of the image better as well. This is useful as it provides us with another dimension of data after we extract the general color characteristics of the image (with 1st order measures) and the texture characteristics of the image (with the GLCM). Template matching allows us to recognize general spatial patterns in the image. Template matching is also sensitive to rotations and deformations in the image. Our dataset also contains several rotated images. This may potentially cause several problems for template matching as it relies on only one image and the transformed images will not conform to this template.

### Feature Preprocessing and Normalization

```python
#Define the exact order of features for the vector
firstOrderNames = ["meanR", "meanG", "meanB", "varR", "varG", "varB"]
glcmNames = list()
for d in distancesGLCM:
    for a in anglesGLCM:
        glcmNames.append(f"corrD{d}A{a}")
tmatchNames = ["tmatchBasil", "tmatchChinar", "tmatchLemon"]

allFeatureNames = firstOrderNames + glcmNames + tmatchNames


#Extract data from featuresDict into rows
xList = list()
yList = list()

for label in featuresDict.keys():
    for imgIdx in range(len(featuresDict[label])):
        featDict = featuresDict[label][imgIdx]
        row = list()

        #Add first-order features
        for name in firstOrderNames:
            row.append(featDict[name])

        #Add GLCM correlation features
        corrMatrix = featDict["correlation"]   # expected shape: (len(distancesGLCM), len(anglesGLCM))
        for dIdx in range(len(distancesGLCM)):
            for aIdx in range(len(anglesGLCM)):
                row.append(corrMatrix[dIdx, aIdx])

        #Add template-matching features
        for name in tmatchNames:
            row.append(featDict[name])

        xList.append(row)
        yList.append(label)


#Convert to numpy arrays
X = np.array(xList, dtype=float)
y = np.array(yList)


#Z-score standardization (StandardScaler)
scaler = StandardScaler()
xStandardized = scaler.fit_transform(X)

dfNonStandardized = pd.DataFrame(X,columns=allFeatureNames)

#Create a DataFrame for easy inspection/reporting
dfFinal = pd.DataFrame(xStandardized, columns=allFeatureNames)
dfFinal['class'] = y

#Create a dataFrame of non-standardized features as well for pairplots
dfNonStandardized['class'] = y

dfFinal.head(len(dfFinal))
```

# Estimate the performance of the models with nested cross-validation

Estimate model performance using nested cross-validation.

### Import necessary modules

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
```

### Performing cross-validation for all models

```python
#Define inner and outer cross-validation settings
innerCV = StratifiedKFold(n_splits=5, shuffle=True)   #Model selection
outerCV = StratifiedKFold(n_splits=4, shuffle=True)   #Performance estimation

#Define the classifiers and their parameters as same ones used in the previous step to then just perform cross validation in one loop
classifiers = {
    "Ridge": {
        "classifier": RidgeClassifier(),
        "paramGrid": {
            "alpha": [0.001, 0.01, 0.1, 1]
        }
    },
    "RandomForest": {
        "classifier": RandomForestClassifier(random_state=42),
        "paramGrid": {
            "n_estimators": [50, 100, 150, 200, 250, 300],
            "max_features": ["sqrt", "log2", None],
            "bootstrap": [True]
        }
    },
    "MLP": {
        "classifier": MLPClassifier(max_iter=100000),
        "paramGrid": {
            "hidden_layer_sizes": [i for i in range(15,45,5)],
            "activation": ['tanh', 'relu'],
            "solver": ["sgd", "adam"],
            "validation_fraction": [0.1, 0.3],
            "alpha": [0.01, 0.1, 1]
        }
    }
}

#Go thru each classifier in the dictionary that was defined
for clfName, clfInfo in classifiers.items():

    #Define the lists to store the necessary stats during the process
    outerScores = list()        
    bestParams = list()         
    allYTrue = list()          
    allYPred = list()          
    bestAccuracy = 0        
    bestModel = None
            
    #Outer loop for performance estimation
    for trainIdx, testIdx in outerCV.split(X, y):
        xTrain, xTest = X[trainIdx], X[testIdx]
        yTrain, yTest = y[trainIdx], y[testIdx]

        #Inner loop for model selection with GridSearchCV
        gridSearch = GridSearchCV(
            estimator=clfInfo["classifier"],
            param_grid=clfInfo["paramGrid"],
            cv=innerCV
        )
        gridSearch.fit(xTrain, yTrain)

        #Evaluate the best model on the outer test fold
        yPred = gridSearch.predict(xTest)
        foldAccuracy = accuracy_score(yTest, yPred)
        outerScores.append(foldAccuracy)
        bestParams.append(gridSearch.best_params_)

        #Collect predictions for the overall confusion matrix
        allYTrue.extend(yTest)
        allYPred.extend(yPred)

        #Track the best model across all outer folds
        if foldAccuracy > bestAccuracy:
            bestAccuracy = foldAccuracy
            bestModel = gridSearch.best_estimator_

    #Report results
    print(f"Classifier: {clfName}")
    print(f"\nBest hyperparameters per outer fold:")
    i = 0
    for params in bestParams:
        print(f"Fold {i+1}: {params}")
        i += 1

    print(f"\nOuter fold accuracies: {[score for score in outerScores]}")
    print(f"Mean accuracy (outer folds): {np.mean(outerScores):.4f}")
    print(f"Best single fold accuracy: {bestAccuracy:.4f}")

    #Plot confusion matrix
    cm = confusion_matrix(allYTrue, allYPred, labels=classes)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classes)
    fig, ax = plt.subplots(figsize=(6, 5))
    disp.plot(ax=ax, colorbar=False)
    ax.set_title(f'Confusion Matrix - {clfName}\n(Mean Accuracy: {np.mean(outerScores):.4f})')
    plt.tight_layout()
    plt.show()
```

The mean accuracy of the outer folds represents the evaluation score of the models on an unseen dataset. This provides us with an unbiased estimate for the performance of the model with the selected hyperparameters. By using a nested structure like this, the model selection and evaluation processes are kept sepearte from each other, hence preventing a biased evaluation of the model since in a single loop cross-validation, we might choose the models that appeal to the specific evaluation folds (in other words, some models can memorize the evaluation fold (leakage of test data to training process), hence leading to overfitting.). However, using a structure like this, it allows to simulate the accuracy much better with how the model would perform with completely unseen data.

#### Best Classifier
We can decide which model performed best by looking at the mean accuracy across all outer folds for each model for the nested cross validation. According to this, the ridge classifier performed the best with a mean accuracy of %94.02. This could be because of the relationships between the features that was observed during plotting section. As can be seen from the pairplots, there are many features that are correlated with each other. Also, if we look at the PCA, it can be seen that the classes were pretty distinct from each other and they can be seperated nicely using linear boundaries. One other reason could also be the size of our dataset. Methods that depend on linear regression such as the ridge classifier usually tend to outperform models like MLP and random forest since these usually require much karger data to perform better.

#### Limitations /Improvements
One big limitation in the project was during the template matching phase. Normally, for a dataset with different rotations and different scales for images, it isn't advisable to choose only a single template image and test the score according to that since the equations and technique used in template matching is sensitive to transformations of the image. Using the PCA technique on template images and then performing template matching can help solve the transformation issue for the template matching. Other approaches were also mentioned in the lecture such as using Fourier and Mellin transforms etc...

Apart from this, the biggest limitation would be the small size of our dataset. Due to this size, we can't really have seperate train/test splits and have to use cross validation while training. Even though the nested cross-validation mostly gets rid of the bias and give us unbiased estimate of the model performence, it would still be much healthy to be able to have seperate splits.

Also, since manual feature extraction were performed and also, very few features were used to train the models, the models may not be able to capture some of the larger or more complex patterns in the images. One possible way to solve this issue could be to use the deep learning based approches such as CNNs however, as mentioned before, the small dataset size would prevent us to be able to healthily train a DL based model. Another improvement to make could be to add even more feature extraction techniques that try to capture more diverse features in the images.
